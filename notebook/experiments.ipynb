## Weather Classification Dataset Creation
This includes every possible pattern that can be there in weather

```python
import pandas as pd
import numpy as np
import random

def generate_weather_data(n_samples=10000, random_seed=42):
    np.random.seed(random_seed)
    weather_conditions = ['Sunny', 'Cloudy', 'Rainy', 'Stormy', 'Snowy', 'Foggy', 'Windy', 'Hail', 'Overcast', 'Drizzle']
    data = []

    for _ in range(n_samples):
        condition = random.choice(weather_conditions)

        if condition == 'Sunny':
            temperature = np.random.uniform(25, 40)
            humidity = np.random.uniform(10, 40)
            pressure = np.random.uniform(1005, 1025)
            wind_speed = np.random.uniform(2, 15)
            wind_direction = np.random.uniform(0, 360)
            precipitation = np.random.uniform(0, 0.5)
            visibility = np.random.uniform(8, 10)
            dew_point = temperature - np.random.uniform(8, 12)
            cloud_cover = np.random.uniform(0, 20)
            solar_radiation = np.random.uniform(600, 1000)
            uv_index = np.random.uniform(6, 10)

        elif condition == 'Cloudy':
            temperature = np.random.uniform(18, 30)
            humidity = np.random.uniform(40, 70)
            pressure = np.random.uniform(1000, 1020)
            wind_speed = np.random.uniform(5, 20)
            wind_direction = np.random.uniform(0, 360)
            precipitation = np.random.uniform(0, 1)
            visibility = np.random.uniform(5, 9)
            dew_point = temperature - np.random.uniform(3, 6)
            cloud_cover = np.random.uniform(60, 90)
            solar_radiation = np.random.uniform(200, 500)
            uv_index = np.random.uniform(2, 5)

        elif condition == 'Rainy':
            temperature = np.random.uniform(15, 28)
            humidity = np.random.uniform(70, 100)
            pressure = np.random.uniform(990, 1010)
            wind_speed = np.random.uniform(10, 30)
            wind_direction = np.random.uniform(0, 360)
            precipitation = np.random.uniform(1, 20)
            visibility = np.random.uniform(2, 8)
            dew_point = temperature - np.random.uniform(1, 3)
            cloud_cover = np.random.uniform(80, 100)
            solar_radiation = np.random.uniform(50, 250)
            uv_index = np.random.uniform(0, 3)

        elif condition == 'Stormy':
            temperature = np.random.uniform(20, 32)
            humidity = np.random.uniform(70, 100)
            pressure = np.random.uniform(970, 990)
            wind_speed = np.random.uniform(30, 100)
            wind_direction = np.random.uniform(0, 360)
            precipitation = np.random.uniform(10, 50)
            visibility = np.random.uniform(0.5, 5)
            dew_point = temperature - np.random.uniform(0, 2)
            cloud_cover = np.random.uniform(90, 100)
            solar_radiation = np.random.uniform(0, 100)
            uv_index = np.random.uniform(0, 1)

        elif condition == 'Snowy':
            temperature = np.random.uniform(-15, 5)
            humidity = np.random.uniform(60, 100)
            pressure = np.random.uniform(990, 1010)
            wind_speed = np.random.uniform(5, 25)
            wind_direction = np.random.uniform(0, 360)
            precipitation = np.random.uniform(2, 30)
            visibility = np.random.uniform(1, 7)
            dew_point = temperature - np.random.uniform(0, 5)
            cloud_cover = np.random.uniform(70, 100)
            solar_radiation = np.random.uniform(0, 200)
            uv_index = np.random.uniform(0, 2)

        elif condition == 'Foggy':
            temperature = np.random.uniform(0, 15)
            humidity = np.random.uniform(90, 100)
            pressure = np.random.uniform(1005, 1020)
            wind_speed = np.random.uniform(0, 10)
            wind_direction = np.random.uniform(0, 360)
            precipitation = np.random.uniform(0, 0.2)
            visibility = np.random.uniform(0.1, 2)
            dew_point = temperature - np.random.uniform(0, 1)
            cloud_cover = np.random.uniform(80, 100)
            solar_radiation = np.random.uniform(0, 100)
            uv_index = np.random.uniform(0, 1)

        elif condition == 'Windy':
            temperature = np.random.uniform(15, 30)
            humidity = np.random.uniform(40, 80)
            pressure = np.random.uniform(995, 1015)
            wind_speed = np.random.uniform(25, 60)
            wind_direction = np.random.uniform(0, 360)
            precipitation = np.random.uniform(0, 2)
            visibility = np.random.uniform(6, 10)
            dew_point = temperature - np.random.uniform(3, 6)
            cloud_cover = np.random.uniform(20, 70)
            solar_radiation = np.random.uniform(300, 700)
            uv_index = np.random.uniform(3, 7)

        elif condition == 'Hail':
            temperature = np.random.uniform(-5, 10)
            humidity = np.random.uniform(70, 100)
            pressure = np.random.uniform(980, 1000)
            wind_speed = np.random.uniform(10, 40)
            wind_direction = np.random.uniform(0, 360)
            precipitation = np.random.uniform(5, 25)
            visibility = np.random.uniform(1, 6)
            dew_point = temperature - np.random.uniform(1, 3)
            cloud_cover = np.random.uniform(90, 100)
            solar_radiation = np.random.uniform(0, 150)
            uv_index = np.random.uniform(0, 2)

        elif condition == 'Overcast':
            temperature = np.random.uniform(10, 25)
            humidity = np.random.uniform(60, 90)
            pressure = np.random.uniform(995, 1015)
            wind_speed = np.random.uniform(5, 20)
            wind_direction = np.random.uniform(0, 360)
            precipitation = np.random.uniform(0, 1)
            visibility = np.random.uniform(4, 9)
            dew_point = temperature - np.random.uniform(2, 5)
            cloud_cover = np.random.uniform(80, 100)
            solar_radiation = np.random.uniform(100, 250)
            uv_index = np.random.uniform(1, 3)

        elif condition == 'Drizzle':
            temperature = np.random.uniform(15, 25)
            humidity = np.random.uniform(80, 100)
            pressure = np.random.uniform(995, 1010)
            wind_speed = np.random.uniform(5, 15)
            wind_direction = np.random.uniform(0, 360)
            precipitation = np.random.uniform(0.5, 3)
            visibility = np.random.uniform(3, 8)
            dew_point = temperature - np.random.uniform(1, 2)
            cloud_cover = np.random.uniform(80, 100)
            solar_radiation = np.random.uniform(100, 300)
            uv_index = np.random.uniform(0, 2)

        data.append([
            temperature, humidity, pressure, wind_speed, wind_direction,
            precipitation, visibility, dew_point, cloud_cover, solar_radiation,
            uv_index, condition
        ])

    columns = [
        'temperature_C', 'humidity_%', 'pressure_hPa', 'wind_speed_kmh',
        'wind_direction_deg', 'precipitation_mm', 'visibility_km', 'dew_point_C',
        'cloud_cover_%', 'solar_radiation_Wm2', 'uv_index', 'weather_condition'
    ]
    df = pd.DataFrame(data, columns=columns)
    return df


# Generate dataset
df_weather = generate_weather_data(n_samples=10000)
df_weather.to_csv("synthetic_weather_dataset.csv", index=False)
print("✅ Synthetic weather dataset created: synthetic_weather_dataset.csv")
print(df_weather.head())
```
